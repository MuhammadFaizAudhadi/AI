## 2. Minimal Theory — Dasar Teori

### Resistansi terukur
Jika beda potensial \\(\\Delta V\\) dan arus listrik \\(I\\) tersedia, resistansi dihitung dengan:

\\[
R = \\frac{\\Delta V}{I}
\\]

Jika alat sudah memberikan nilai resistansi, nilai tersebut dapat langsung digunakan setelah satuan dan kualitas pengukuran diperiksa.

### Resistivitas semu
Resistivitas semu dihitung dengan:

\\[
\\rho_a = K R
\\]

Untuk konfigurasi Schlumberger, dengan \(AB/2=a\) dan \(MN/2=b\), faktor geometri adalah:

\\[
K = \\frac{\\pi(a^2-b^2)}{2b}
\\]

Keterangan:
- \\(\\rho_a\\): resistivitas semu (\\(\\Omega\\cdot m\\)).
- \\(K\\): faktor geometri (m).
- \\(R\\): resistansi (\\(\\Omega\\)).
- \\(AB/2\\): setengah jarak elektroda arus (m).
- \\(MN/2\\): setengah jarak elektroda potensial (m).

Rumus tersebut berlaku untuk susunan Schlumberger simetris dengan \(AB/2 > MN/2 > 0\). Pastikan definisi jarak pada catatan lapangan sesuai dengan rumus ini.

### Interpretasi awal
- Nilai resistivitas rendah dapat berkaitan dengan material lempungan, tanah jenuh air, atau fluida yang lebih konduktif.
- Nilai resistivitas tinggi dapat berkaitan dengan batuan kompak, material kering, atau batuan dengan porositas rendah.
- Rentang resistivitas berbagai litologi saling tumpang tindih. Karena itu, resistivitas tidak boleh digunakan sendiri untuk memastikan jenis batuan, keberadaan air tanah, atau kondisi geologi tertentu.

## Rancangan Workflow Python

3. **Perhitungan** — hitung faktor geometri Schlumberger dan resistivitas semu.

**Catatan:** Data di sel kode berikut adalah contoh sintetis untuk menunjukkan cara kerja kode. Ganti dengan data lapangan sebelum melakukan interpretasi penelitian.


```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from pathlib import Path

# ==========================================================
# A. INPUT DATA
# ==========================================================
# GANTI data contoh ini dengan data pengukuran lapangan.
# Semua jarak menggunakan meter dan resistansi menggunakan ohm.
data = pd.DataFrame({
    "titik": ["VES-01"] * 8,
    "AB2_m": [1.0, 1.5, 2.0, 3.0, 4.0, 6.0, 8.0, 10.0],
    "MN2_m": [0.2, 0.2, 0.2, 0.5, 0.5, 0.5, 1.0, 1.0],
    "R_ohm": [18.0, 12.5, 9.2, 6.4, 4.8, 3.1, 2.6, 2.1]
})

data
```

```python
# ==========================================================
# B. QA/QC DATA
# ==========================================================
required = ["titik", "AB2_m", "MN2_m", "R_ohm"]
missing_cols = [col for col in required if col not in data.columns]
if missing_cols:
    raise ValueError(f"Kolom wajib tidak tersedia: {missing_cols}")

for col in ["AB2_m", "MN2_m", "R_ohm"]:
    data[col] = pd.to_numeric(data[col], errors="coerce")

if data[["AB2_m", "MN2_m", "R_ohm"]].isna().any().any():
    raise ValueError("Ada nilai kosong atau nonnumerik pada kolom jarak/resistansi.")

if (data["AB2_m"] <= 0).any() or (data["MN2_m"] <= 0).any():
    raise ValueError("AB/2 dan MN/2 harus lebih besar dari nol.")

if (data["AB2_m"] <= data["MN2_m"]).any():
    raise ValueError("Untuk rumus ini, AB/2 harus lebih besar daripada MN/2.")

if (data["R_ohm"] <= 0).any():
    raise ValueError("Resistansi harus positif. Periksa data dan satuannya.")

print(f"Jumlah data: {len(data)}")
print("Pemeriksaan dasar selesai. Tinjau juga catatan lapangan dan satuan instrumen.")
```

```python
# ==========================================================
# C. HITUNG FAKTOR GEOMETRI DAN RESISTIVITAS SEMU
# ==========================================================
a = data["AB2_m"].to_numpy(dtype=float)
b = data["MN2_m"].to_numpy(dtype=float)
R = data["R_ohm"].to_numpy(dtype=float)

K = np.pi * (a**2 - b**2) / (2 * b)
data["K_m"] = K
data["rho_a_ohm_m"] = K * R

display(data)
```

```python
# ==========================================================
# E. RINGKASAN DAN EKSPOR HASIL
# ==========================================================
print("Rentang AB/2 (m):", data["AB2_m"].min(), "–", data["AB2_m"].max())
print("Rentang resistivitas semu (ohm m):",
      round(data["rho_a_ohm_m"].min(), 3), "–",
      round(data["rho_a_ohm_m"].max(), 3))

# Simpan hasil pada folder kerja notebook.
output_file = Path("hasil_resistivitas_semu.csv")
data.to_csv(output_file, index=False)
print("Tabel hasil disimpan ke:", output_file.resolve())

print(
    "\nInterpretasi awal: amati perubahan tren kurva resistivitas semu "
    "terhadap AB/2. Jangan mengubah tren ini langsung menjadi kedalaman "
    "atau litologi tanpa inversi dan informasi geologi pendukung."
)
```
