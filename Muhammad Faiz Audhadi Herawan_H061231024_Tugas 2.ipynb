### Visualisasi Data ERT — Penampang Resistivitas

Fungsi berikut membuat gambar sebaran resistivitas berdasarkan kolom posisi, kedalaman, dan resistivitas. Sumbu kedalaman dibalik agar nilai kedalaman bertambah ke bawah. Grafik ini merupakan visualisasi titik data, bukan pengganti penampang hasil inversi ERT 2D. Sesuaikan nama kolom dengan CSV yang digunakan.


```python

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy.optimize import least_squares

# Mengubah seed agar variasi data/noise acak ikut berubah
np.random.seed(123)

# ------------------------------------------------------------
# 1. PARAMETER SURVEI DAN MODEL SIMULASI (DATA DIUBAH)
# ------------------------------------------------------------

I_mA = 150.0                # Arus dalam mA (diubah dari 100 mA)
I = I_mA / 1000.0           # Konversi ke ampere

# Setengah jarak elektroda arus AB/2 (m) - Diperluas
AB2 = np.array([1, 2, 3, 5, 7, 10, 15, 20, 25, 35, 50],
               dtype=float)

# Setengah jarak elektroda potensial MN/2 (m)
MN2 = np.where(AB2 <= 10, 0.5, 1.0)

# Parameter model simulasi yang DIUBAH (Model baru: Lapisan atas sedang, bawah konduktif, basement resistif)
rho_model = np.array([150.0, 35.0, 300.0])  # ohm m
h_model = np.array([4.0, 8.0])              # meter

print("MODEL SIMULASI BARU")
print(f"Resistivitas lapisan 1 = {rho_model[0]} ohm m")
print(f"Resistivitas lapisan 2 = {rho_model[1]} ohm m")
print(f"Resistivitas lapisan 3 = {rho_model[2]} ohm m")
print(f"Ketebalan lapisan 1 = {h_model[0]} m")
print(f"Ketebalan lapisan 2 = {h_model[1]} m")

# ------------------------------------------------------------
# 2. FAKTOR GEOMETRI SCHLUMBERGER
# ------------------------------------------------------------

AB = 2 * AB2
MN = 2 * MN2

K = np.pi * (AB**2 - MN**2) / (4 * MN)

# ------------------------------------------------------------
# 3. MODEL RESPONS SIMULASI
# ------------------------------------------------------------
def respons_aproksimasi(log_ab2, params):
    rho1, rho2, rho3 = 10 ** params[:3]
    h1, h2 = 10 ** params[3:]

    z1 = h1
    z2 = h1 + h2

    q1 = 1 / (1 + np.exp(-4 * (log_ab2 - np.log10(z1 + 0.5))))
    q2 = 1 / (1 + np.exp(-4 * (log_ab2 - np.log10(z2 + 0.5))))

    w1 = 1 - q1
    w2 = q1 * (1 - q2)
    w3 = q2

    log_rhoa = (
        w1 * np.log(rho1)
        + w2 * np.log(rho2)
        + w3 * np.log(rho3)
    )

    return np.exp(log_rhoa)


# ------------------------------------------------------------
# 4. MEMBUAT DATA RESISTIVITAS SEMU SIMULASI
# ------------------------------------------------------------
parameter_sebenarnya = np.array([
    np.log10(rho_model[0]),
    np.log10(rho_model[1]),
    np.log10(rho_model[2]),
    np.log10(h_model[0]),
    np.log10(h_model[1])
])

rho_a_teori = respons_aproksimasi(
    np.log10(AB2), parameter_sebenarnya
)

# Noise 2.5% untuk variasi data
rho_a_obs = rho_a_teori * (
    1 + np.random.normal(0, 0.025, len(AB2))
)

deltaV = rho_a_obs * I / K
deltaV_mV = deltaV * 1000

df = pd.DataFrame({
    "AB/2_m": AB2,
    "MN/2_m": MN2,
    "AB_m": AB,
    "MN_m": MN,
    "Arus_mA": I_mA,
    "Beda_potensial_mV": deltaV_mV,
    "Faktor_geometri_m": K,
    "Resistivitas_semu_ohm_m": K * (deltaV / I)
})

print("\nTABEL DATA GEOLISTRIK SIMULASI (BARU)")
print(df.round(3).to_string(index=False))

df.to_csv("data_geolistrik_simulasi_baru.csv", index=False)
print("\nFile disimpan: data_geolistrik_simulasi_baru.csv")

# ------------------------------------------------------------
# 5. INVERSI 1D SEDERHANA MENGGUNAKAN SCIPY
# ------------------------------------------------------------
def residual(params):
    prediksi = respons_aproksimasi(np.log10(AB2), params)
    return np.log(prediksi) - np.log(rho_a_obs)

tebakan_awal = np.array([
    np.log10(100.0),
    np.log10(30.0),
    np.log10(250.0),
    np.log10(3.0),
    np.log10(6.0)
])

batas_bawah = np.array([
    np.log10(1), np.log10(1), np.log10(1),
    np.log10(0.2), np.log10(0.2)
])

batas_atas = np.array([
    np.log10(2000), np.log10(2000), np.log10(2000),
    np.log10(50), np.log10(100)
])

hasil_inv = least_squares(
    residual,
    tebakan_awal,
    bounds=(batas_bawah, batas_atas),
    max_nfev=10000
)

parameter_inv = 10 ** hasil_inv.x
rho_inv = parameter_inv[:3]
h_inv = parameter_inv[3:]

rho_pred = respons_aproksimasi(
    np.log10(AB2), hasil_inv.x
)

rms_log = np.sqrt(np.mean(
    (np.log(rho_pred) - np.log(rho_a_obs))**2
))
misfit_persen = 100 * np.sqrt(np.mean(
    ((rho_pred - rho_a_obs) / rho_a_obs)**2
))

hasil_lapisan = pd.DataFrame({
    "Lapisan": [1, 2, 3],
    "Resistivitas_inversi_ohm_m": rho_inv,
    "Ketebalan_m": [h_inv[0], h_inv[1], np.nan],
    "Kedalaman_batas_bawah_m": [
        h_inv[0], h_inv[0] + h_inv[1], np.nan
    ]
})

print("\nHASIL INVERSI 1D (BARU)")
print(hasil_lapisan.round(3).to_string(index=False))
print(f"\nRMS log residual = {rms_log:.4f}")
print(f"Misfit relatif RMS = {misfit_persen:.2f}%")

# ------------------------------------------------------------
# 6. GRAFIK KURVA RESISTIVITAS SEMU
# ------------------------------------------------------------
ab_smooth = np.logspace(
    np.log10(AB2.min()),
    np.log10(AB2.max()),
    250
)

rho_smooth = respons_aproksimasi(
    np.log10(ab_smooth), hasil_inv.x
)

plt.figure(figsize=(9, 6))
plt.loglog(AB2, rho_a_obs, "o", label="Data simulasi baru", markersize=6)
plt.loglog(AB2, rho_pred, "s", label="Respons model inversi", markersize=5)
plt.loglog(ab_smooth, rho_smooth, "-", label="Kurva model inversi")
plt.xlabel("AB/2 (m)")
plt.ylabel("Resistivitas semu (ohm m)")
plt.title("Kurva Resistivitas Semu - Schlumberger (Data Baru)")
plt.grid(True, which="both", alpha=0.3)
plt.legend()
plt.tight_layout()
plt.show()

# ------------------------------------------------------------
# 7. MODEL PENAMPANG 2D KONSEPTUAL (WARNA DIUBAH MENJADI 'viridis')
# ------------------------------------------------------------
x = np.linspace(0, 60, 200)
z_max = max(30, h_inv.sum() + 10)
z = np.linspace(0, z_max, 150)

rho_section = np.zeros((len(z), len(x)))
z1 = h_inv[0]
z2 = h_inv[0] + h_inv[1]

for iz, depth in enumerate(z):
    if depth < z1:
        rho_section[iz, :] = rho_inv[0]
    elif depth < z2:
        rho_section[iz, :] = rho_inv[1]
    else:
        rho_section[iz, :] = rho_inv[2]

plt.figure(figsize=(11, 5))
mesh = plt.pcolormesh(
    x, z, rho_section,
    shading="auto", 
    cmap="viridis",  # Warna diubah dari 'turbo' menjadi 'viridis' (bisa diganti 'plasma', 'jet', dll)
    norm=plt.matplotlib.colors.LogNorm(
        vmin=max(1, rho_inv.min()),
        vmax=rho_inv.max()
    )
)

plt.gca().invert_yaxis()
plt.axhline(z1, color="cyan", linestyle="--", label=f"Batas lapisan 1-2: {z1:.1f} m")
plt.axhline(z2, color="orange", linestyle="--", label=f"Batas lapisan 2-3: {z2:.1f} m")
plt.xlabel("Jarak lintasan (m)")
plt.ylabel("Kedalaman (m)")
plt.title("Penampang Resistivitas 2D Konseptual (Colormap: Viridis)")
plt.colorbar(mesh, label="Resistivitas (ohm m)")
plt.legend(loc="upper right")
plt.tight_layout()
plt.show()

# ------------------------------------------------------------
# 8. INTERPRETASI AWAL
# ------------------------------------------------------------
def interpretasi(rho):
    if rho < 10:
        return "Sangat konduktif; lempung atau fluida asin mungkin"
    elif rho < 30:
        return "Konduktif; lempung atau sedimen jenuh air mungkin"
    elif rho < 100:
        return "Resistivitas menengah; sedimen atau material lapuk"
    elif rho < 300:
        return "Relatif resistif; pasir, kerikil, atau batuan mungkin"
    else:
        return "Resistivitas tinggi; batuan kompak atau material kering mungkin"

print("\nINTERPRETASI AWAL MODEL BARU")
for i, rho in enumerate(rho_inv, start=1):
    print(f"Lapisan {i}: {rho:.2f} ohm m -> {interpretasi(rho)}")
```

### Visualisasi Kurva HVSR

Fungsi ini menggambar kurva H/V terhadap frekuensi dan menandai puncak maksimum yang digunakan sebagai identifikasi awal f₀ dan A₀. Periode dominan T₀ dihitung sebagai 1/f₀. Pastikan kurva sudah melalui pemrosesan dan pemeriksaan kualitas HVSR yang sesuai.


```python
# ============================================================
# PENGOLAHAN DATA MIKROTREMOR - METODE HVSR
# Visualisasi Kurva H/V, Frekuensi Dominan (f₀), & Periode (T₀)
# ============================================================

# Memastikan gambar/grafik dirender langsung di dalam Jupyter Notebook
%matplotlib inline

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

# ------------------------------------------------------------
# 1. MEMBUAT DATA SINTETIS KURVA HVSR (CONTOH)
# ------------------------------------------------------------
np.random.seed(42)
frekuensi_sintetis = np.logspace(np.log10(0.2), np.log10(20), 200)

# Simulasi puncak HVSR pada f₀ = 2.5 Hz dengan amplitudo A₀ = 4.2
f0_target = 2.5
h_v_sintetis = 1.0 + 3.2 * np.exp(-((np.log(frekuensi_sintetis) - np.log(f0_target))**2) / 0.15)
h_v_sintetis += np.random.normal(0, 0.08, len(frekuensi_sintetis)) # Penambahan sedikit noise
h_v_sintetis = np.clip(h_v_sintetis, 0.5, None)

data_hvsr = pd.DataFrame({
    "frequency": frekuensi_sintetis,
    "hv": h_v_sintetis
})

# Menyimpan data contoh ke file CSV
data_hvsr.to_csv("kurva_hvsr_simulasi.csv", index=False)
print("File contoh 'kurva_hvsr_simulasi.csv' berhasil dibuat.")

# ------------------------------------------------------------
# 2. FUNGSI ANALISIS KURVA HVSR
# ------------------------------------------------------------
def analisis_hvsr(
    df,
    kolom_frekuensi="frequency",
    kolom_hv="hv",
    f_min=None,
    f_max=None,
    informasi_geologi=None
):
    """Menganalisis kurva HVSR untuk menentukan f₀, A₀, dan T₀."""
    for kolom in (kolom_frekuensi, kolom_hv):
        if kolom not in df.columns:
            raise ValueError(
                f"Kolom '{kolom}' tidak ditemukan. Kolom tersedia: {list(df.columns)}"
            )

    data = df[[kolom_frekuensi, kolom_hv]].copy()
    data[kolom_frekuensi] = pd.to_numeric(data[kolom_frekuensi], errors="coerce")
    data[kolom_hv] = pd.to_numeric(data[kolom_hv], errors="coerce")
    data = data.replace([np.inf, -np.inf], np.nan).dropna()
    data = data[(data[kolom_frekuensi] > 0) & (data[kolom_hv] > 0)]

    if f_min is not None:
        data = data[data[kolom_frekuensi] >= f_min]
    if f_max is not None:
        data = data[data[kolom_frekuensi] <= f_max]
    if data.empty:
        raise ValueError("Tidak ada data valid pada rentang frekuensi yang dipilih.")

    puncak = data.loc[data[kolom_hv].idxmax()]
    f0 = float(puncak[kolom_frekuensi])
    a0 = float(puncak[kolom_hv])
    t0 = 1.0 / f0

    narasi = (
        f"Analisis kurva HVSR menggunakan {len(data)} pasangan data frekuensi dan amplitudo H/V. "
        f"Puncak maksimum kurva berada pada frekuensi dominan (f₀) sebesar {f0:.4g} Hz, "
        f"dengan amplitudo H/V (A₀) sebesar {a0:.4g}. Periode dominan dihitung menggunakan T₀ = 1/f₀, "
        f"sehingga diperoleh T₀ = {t0:.4g} detik."
    )
    return narasi, {"f0_Hz": f0, "A0_HV": a0, "T0_s": t0}

# ------------------------------------------------------------
# 3. FUNGSI VISUALISASI KURVA HVSR (MEMUNCULKAN GAMBAR)
# ------------------------------------------------------------
def plot_kurva_hvsr(
    df,
    kolom_frekuensi="frequency",
    kolom_hv="hv",
    f_min=None,
    f_max=None,
    simpan_ke=None
):
    """Plot kurva HVSR dan menandai puncak maksimum agar gambar muncul."""
    for kolom in (kolom_frekuensi, kolom_hv):
        if kolom not in df.columns:
            raise ValueError(
                f"Kolom '{kolom}' tidak ditemukan. Kolom tersedia: {list(df.columns)}"
            )

    data = df[[kolom_frekuensi, kolom_hv]].copy()
    data[kolom_frekuensi] = pd.to_numeric(data[kolom_frekuensi], errors="coerce")
    data[kolom_hv] = pd.to_numeric(data[kolom_hv], errors="coerce")
    data = data.replace([np.inf, -np.inf], np.nan).dropna()
    data = data[(data[kolom_frekuensi] > 0) & (data[kolom_hv] > 0)]
    
    if f_min is not None:
        data = data[data[kolom_frekuensi] >= f_min]
    if f_max is not None:
        data = data[data[kolom_frekuensi] <= f_max]
    if data.empty:
        raise ValueError("Tidak ada data HVSR valid pada rentang frekuensi yang dipilih.")

    data = data.sort_values(kolom_frekuensi)
    puncak = data.loc[data[kolom_hv].idxmax()]
    f0 = float(puncak[kolom_frekuensi])
    a0 = float(puncak[kolom_hv])
    t0 = 1.0 / f0

    # Membuat plot matplotlib
    fig, ax = plt.subplots(figsize=(10, 5))
    ax.plot(data[kolom_frekuensi], data[kolom_hv], linewidth=1.6, label="Kurva H/V", color="blue")
    ax.scatter([f0], [a0], s=70, zorder=3, color="red", label=f"Puncak: f₀={f0:.4g} Hz; A₀={a0:.4g}")
    ax.axvline(f0, color="gray", linestyle="--", linewidth=1, label=f"T₀={t0:.4g} s")
    
    ax.set_xlabel("Frekuensi (Hz)")
    ax.set_ylabel("Amplitudo H/V")
    ax.set_title("Kurva Horizontal-to-Vertical Spectral Ratio (HVSR)")
    ax.grid(alpha=0.25)
    ax.legend()
    fig.tight_layout()
    
    if simpan_ke:
        fig.savefig(simpan_ke, dpi=300, bbox_inches="tight")
        
    plt.show()  # Perintah utama untuk menampilkan gambar di notebook
    return fig, ax, {"f0_Hz": f0, "A0_HV": a0, "T0_s": t0}

# ------------------------------------------------------------
# 4. EKsekusi PEMANGGILAN FUNGSI (AKTIF / TANPA KOMENTAR)
# ------------------------------------------------------------
print("\n--- HASIL ANALISIS NARASI HVSR ---")
narasi_hasil, parameter_hasil = analisis_hvsr(
    data_hvsr, 
    kolom_frekuensi="frequency", 
    kolom_hv="hv"
)
print(narasi_hasil)

print("\n--- PARAMETER TERHITUNG ---")
print(parameter_hasil)

print("\n--- MENAMPILKAN GRAFIK KURVA HVSR ---")
fig, ax, parameter = plot_kurva_hvsr(
    data_hvsr, 
    kolom_frekuensi="frequency", 
    kolom_hv="hv",
    simpan_ke="kurva_HVSR.png"
)
```
